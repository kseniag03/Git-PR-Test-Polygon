```python
import os
import requests
import json
from dotenv import load_dotenv
```

Для выполнения локально необходимо создать .env рядом с ноутбуком и заполнить поле VT_API_KEY (ввести ключ VirusTotal API).

```python
load_dotenv()

API_KEY = os.getenv("VT_API_KEY")

if not API_KEY:
    raise RuntimeError("Не найден VT_API_KEY. Нужно задать переменную окружения.")
```

## Этап 2. Авторизация и запрос

```python
def vt_get_ip_report(ip: str, api_key: str) -> dict:
    url = f"https://www.virustotal.com/api/v3/ip_addresses/{ip}"
    headers = {"x-apikey": api_key}
    r = requests.get(url, headers=headers, timeout=30)
    r.raise_for_status()
    
    return r.json()
```

```python
ip = "8.8.8.8"
data = vt_get_ip_report(ip, API_KEY)

print(json.dumps(data, indent=2, ensure_ascii=False)[:4000])
```

## Этап 3. Вывод результатов

```python
def print_ip_stats(data):
    stats = data["data"]["attributes"]["last_analysis_stats"]

    print("IP:", data["data"]["id"])
    print("Reputation:", data["data"]["attributes"].get("reputation"))
    print("Last analysis stats:", stats)

    print("Risk score (malicious+suspicious):", stats.get("malicious", 0) + stats.get("suspicious", 0))
```

```python
print_ip_stats(data)
```

```python
print_ip_stats(vt_get_ip_report("185.220.101.1", API_KEY))
```
